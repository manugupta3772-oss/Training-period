```python
# Ans. 24)
runs = float(input("Enter runs scored: "))
balls = float(input("Enter balls faced: "))
if balls == 0:
    print("Cannot calculate strike rate because balls faced is zero")
else:
    sr = (runs / balls) * 100
    print("Strike Rate:", round(sr, 2))
    if sr > 150:
        print("Explosive (T20 finisher)")
    elif sr > 100:
        print("Aggressive (Top order T20)")
    elif sr >= 75:
        print("Steady (ODI anchor)")
    else:
        print("Defensive (Test match)")
```

```python
# Ans 34.)

num = int(input("Enter a number: "))
original = num
reverse = 0
if num < 0:
    print("Not a palindrome")
else:
    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10
    print("Reversed number:", reverse)
    if original == reverse:
        print("Palindrome")
    else:
        print("Not a palindrome")
```

```python
# Ans 35.)
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
x = a
y = b
# Calculate HCF using Euclidean algorithm
while y != 0:
    remainder = x % y
    x = y
    y = remainder
hcf = x
# Calculate LCM
lcm = (a * b) // hcf
print("HCF =", hcf)
print("LCM =", lcm)
```

```python
# Ans 37.)
marks = list(map(int, input("Enter 10 marks: ").split()))
maximum = marks[0]
minimum = marks[0]
total = 0
# Find maximum, minimum and total
for m in marks:
    if m > maximum:
        maximum = m
    if m < minimum:
        minimum = m
    total += m
# Calculate average
average = total / len(marks)
# Count students above average
above = 0
for m in marks:
    if m > average:
        above += 1
# Find second highest distinct mark
second = None
for m in marks:
    if m < maximum:
        if second is None or m > second:
            second = m
print("Maximum marks:", maximum)
print("Minimum marks:", minimum)
print("Average marks:", average)
print("Students above average:", above)
print("Second highest marks:", second)
```

```python
# Ans 39.)
states = {
    "Karnataka": "Bengaluru",
    "Maharashtra": "Mumbai",
    "Tamil Nadu": "Chennai",
    "Kerala": "Thiruvananthapuram",
    "Rajasthan": "Jaipur"
}
while True:
    print("\n----- STATE MANAGEMENT -----")
    print("1. Add State")
    print("2. Update State")
    print("3. Delete State")
    print("4. Search State")
    print("5. Display All States")
    print("6. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        state = input("Enter state name: ")
        capital = input("Enter capital: ")
        states[state] = capital
        print("State added successfully!")
    elif choice == 2:
        state = input("Enter state to update: ")
        if state in states:
            capital = input("Enter new capital: ")
            states[state] = capital
            print("State updated successfully!")
        else:
            print("State not found!")
    elif choice == 3:
        state = input("Enter state to delete: ")
        if state in states:
            del states[state]
            print("State deleted successfully!")
        else:
            print("State not found!")
    elif choice == 4:
        state = input("Enter state to search: ")
        if state in states:
            print("Capital:", states[state])
        else:
            print("State not found!")
    elif choice == 5:
        for state in sorted(states):
            print(state, ":", states[state])
    elif choice == 6:
        print("Program exited!")
        break
    else:
        print("Invalid choice!")
```

```python
# Ans 40.)
A = {"Python", "Java", "C++", "SQL"}
B = {"Java", "JavaScript", "Python", "Go", "SQL"}
# 1. Common elements
print("Common elements:", A & B)
# 2. Elements only in A
print("Only in A:", A - B)
# 3. Elements only in B
print("Only in B:", B - A)
# 4. All elements (Union)
print("Union:", A | B)
# 5. Elements present in either set, but not both
print("Symmetric Difference:", A ^ B)
```
